```python
import json
import random

def readjson(file_name):
    with open(file_name, encoding='utf-8') as f:
        data = json.load(f)
    return data

def read_jsonl(file_path):
    data = []
    with open(file_path, 'r', encoding='utf-8') as file:
        for line in file:
            json_obj = json.loads(line)
            data.append(json_obj)
    return data

def question_process(fpath):
    if fpath.endswith('jsonl'):
        data = read_jsonl(fpath)
    else:
        data = readjson(fpath)
    return data

input_file = "./cwq_test_origin_with_topic_alias.json"
test_dataset = question_process(input_file)
data_for_abs = random.sample(test_dataset, 1000)

with open("cwq_test_for_abs.json", "w") as f:
    json.dump(data_for_abs, f)
```
